```python
# --- Task 1: Create Variables

# --- Patient information

patient_name = "Sarah Johnson"
patient_id = "PT001"
patient_age = 42

# --- Appointment information
department = "Cardiology"
doctor_name = "Dr. Musa"
appointment_day = "Monday"
appointment_time ="10:00am"

# --- Emergency status
emergency_case = "False"


# Task 2: Validate Patient Information

# Check if the patient name is not empty
name_valid = patient_name != ""

# Check if the patient ID is not empty
patient_id_valid = patient_id != ""

# Check if the patient age is greater than 0 and not more than 120
age_valid = patient_age > 0 and patient_age <= 120

# Display validation results
print("Name valid:", name_valid)
print("Patient ID valid:", patient_id_valid)
print("Age valid:", age_valid)

# Task 3: Validate Department

# Check if the department is available
if department == "Cardiology":
    print("Department Available")

elif department == "Pediatrics":
    print("Department Available")

elif department == "General Medicine":
    print("Department Available")

elif department == "Orthopedics":
    print("Department Available")

elif department == "Neurology":
    print("Department Available")

else:
    print("Invalid Department")



# Task 4: Categorize Patients

# Determine the patient's age category
if patient_age <= 12:
    patient_category = "Child"

elif patient_age <= 17:
    patient_category = "Teenager"

elif patient_age <= 64:
    patient_category = "Adult"

else:
    patient_category = "Senior"

# Display the patient's category
print("Patient Category:", patient_category)


# Task 5: Determine Appointment Priority

# Check if the patient is an emergency case
if emergency_case:
    priority = "High Priority"

# Check if the patient is a senior
elif patient_category == "Senior":
    priority = "Medium Priority"

# All other patients have normal priority
else:
    priority = "Normal Priority"

# Display the priority level
print("Priority Level:", priority)


# Task 6: Create Validation Functions

# Check if the patient's age is valid
def validate_age(age):
    return age > 0 and age <= 120


# Check if the patient ID is not empty
def validate_patient_id(patient_id):
    return patient_id != ""


# Check if the department is valid
def validate_department(department):
    if department == "Cardiology":
        return True

    elif department == "Pediatrics":
        return True

    elif department == "General Medicine":
        return True

    elif department == "Orthopedics":
        return True

    elif department == "Neurology":
        return True

    else:
        return False

# Task 7: Create an Appointment Function

def create_appointment(name, patient_id, age, department,
                       doctor_name, appointment_day, appointment_time):

    # Validate the appointment information
    id_valid = validate_patient_id(patient_id)
    age_valid = validate_age(age)
    department_valid = validate_department(department)

    # Check if all required information is valid
    if name != "" and id_valid and age_valid and department_valid:

        # Categorize the patient
        if age <= 12:
            category = "Child"

        elif age <= 17:
            category = "Teenager"

        elif age <= 64:
            category = "Adult"

        else:
            category = "Senior"

        # Determine appointment priority
        if category == "Senior":
            priority = "Medium Priority"

        else:
            priority = "Normal Priority"

        # Return appointment details
        return name, patient_id, age, category, department, doctor_name, appointment_day, appointment_time, priority

    else:
        return "Invalid appointment information"

# Task 8: Use a For Loop

# Store multiple patient records
patients = [
    ["Sarah Johnson", 42, "Cardiology"],
    ["James Bello", 70, "Neurology"],
    ["Ada Obi", 8, "Paediatrics"],
    ["John Smith", 45, "Surgery"]
]

# Process each patient
for patient in patients:

    # Get the patient's information
    name = patient[0]
    age = patient[1]
    department = patient[2]

    # Validate the department
    department_valid = validate_department(department)

    # Categorise the patient based on age
    if age <= 12:
        category = "Child"

    elif age <= 17:
        category = "Teenager"

    elif age <= 64:
        category = "Adult"

    else:
        category = "Senior"

    # Determine appointment priority
    if category == "Senior":
        priority = "Medium Priority"

    else:
        priority = "Normal Priority"

    # Display patient information
    print("Patient Name:", name)
    print("Department Valid:", department_valid)
    print("Patient Category:", category)
    print("Priority:", priority)
    print("----------------------------")

# Task 9: Use a While Loop

# Ask the user to enter a patient ID
patient_id = input("Enter Patient ID: ")

# Keep asking while the patient ID is empty
while patient_id == "":
    print("Patient ID cannot be empty.")
    patient_id = input("Enter Patient ID: ")

print("Valid Patient ID:", patient_id)


# Task 10: Infinite While Loop

while True:
    answer = input("Do you want to register another appointment? (Yes/No): ")

    if answer == "Yes":
        print("Registering another appointment...")

    elif answer == "No":
        print("Appointment registration ended.")
        break

    else:
        print("Please enter Yes or No.")

# Task 11: Understand Variable Scope

# Global variable
hospital_name = "SmartCare Hospital"


def show_hospital_details():

    # Local variable
    department = "Cardiology"

    # Use the global variable
    print("Hospital Name:", hospital_name)

    # Use the local variable
    print("Department:", department)


# Call the function
show_hospital_details()


# Task 12: Generate an Appointment Report

print("====================================")
print("   HOSPITAL APPOINTMENT REPORT")
print("====================================")

print("Patient Name:", patient_name)
print("Patient ID:", patient_id)
print("Age:", patient_age)
print("Category:", patient_category)
print("Department:", department)
print("Doctor:", doctor_name)
print("Appointment Day:", appointment_day)
print("Appointment Time:", appointment_time)
print("Emergency Case:", emergency_case)
print("Priority Level:", priority)
print("Appointment Status: Confirmed")

print("====================================")
```
